# Transformación

## Encoding

Aplicar one hot encoding a las variables categóricas propiedad_vivienda y intencion_prestamo. 

```python
cols_ohe = ['propiedad_vivienda', 'intencion_prestamo']
```

```python
from sklearn.preprocessing import OneHotEncoder

ohe = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
encode = ohe.fit_transform(train[cols_ohe])

# Eliminar originales y unir las codificadas
train = train.drop(columns=cols_ohe)
train = pd.concat([train, pd.DataFrame(encode, columns=ohe.get_feature_names_out(cols_ohe), index=train.index)], axis=1)

```

Aplicar ordial encoding a la variable categórica calificacion_prestamo. 

```python
from sklearn.preprocessing import OrdinalEncoder

categorias = [['G', 'F', 'E', 'D', 'C', 'B', 'A']]

oe = OrdinalEncoder(categories=categorias)
train[["calificacion_prestamo"]] = oe.fit_transform(train[["calificacion_prestamo"]])
```

```python
train[["calificacion_prestamo"]]
```

Aplicar lavel encoding a la variable categórica incumplimiento_historial. 

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
train["incumplimiento_historial"] = le.fit_transform(train["incumplimiento_historial"])
```

## Reescalado

Reescalado de las variables fuera de rango [0, 1]

```python
from sklearn.preprocessing import MinMaxScaler

cols_escalar = ["edad", "ingreso", "duracion_empleo", "monto_prestamo", 
                "tasa_interes", "duracion_credito", "calificacion_prestamo"]

scaler = MinMaxScaler()

train[cols_escalar] = scaler.fit_transform(train[cols_escalar])

```
